## Input & output

Datatypes?
- string --> " " / ' ' / """ """ / ''' ''' --> pure tekst
- int --> geheel getal
- float --> kommagetallen
- bool --> boolean --> True en False

```python
#input vragen aan iets of iemand?
#mini opdracht: zoek het datatype van een bepaald getal
x = input()
print(x)
print(f"Het gegeven getal is: {x}")
print("Het gegeven getal is:", x)
print("Het gegeven getal is:" + x)

print(f"Het datatype van x is nu {type(x)}")

#converten van het ene datatype naar het andere
#van str naar int
xAlsInt = int(x)
print(type(xAlsInt))
print(xAlsInt)

xAlsFloat = float(x)
print(type(xAlsFloat))
print(xAlsFloat)

xAlsString = str(xAlsFloat)
print(xAlsString)

```

## Concatenatie en combineren van verschillende outputs
- '+' --> combineren van 2 STRINGS of het optellen van 2 getallen, je moet zelf een spatie vorzien als het een string is en alles blijft een string.
- ',' --> voegt automatisch zelf een spatie toe en laat je toe meerdere datatypes te combineren.

```python
getal = 5
print("Het getal is", getal)
print(getal + getal)
print("Het getal is " + str(getal))

print(str(getal) + str(getal)) # "5" + "5"
print(str(getal + getal))

#print(str(getal) + getal) --> strings en getallen niet op te tellen
```

## Expressions
### Wiskundige operatoren

```python
x = 15
y = 8
print(x,y)

x, y = 10,3
print(x,y)

print(x + y) #13
print(x - y) #7
print(x * y) #30
print(x / y) #3.333333
print(x // y) #gehele deling --> hoeveel keer past y effectief in x --> 3
print(x % y) #1 --> rest na gehele deling
print(x ** y) #1000 --> x tot de y'de macht
```

### Toekenningsoperatoren

```python
x,y = 10,3
x += y; print(x, type(x)) #13 int
x -= y; print(x, type(x)) #10 int
x *= y; print(x, type(x)) #30 int
x /= y; print(x, type(x)) #10.0 float
x //= y; print(x, type(x)) #3.0 float

x = 10
x **= y; print(x, type(x)) #1000 int

x += 1
print(x)
```

### Vergelijkingsoperatoren

```python
x, y = 10, 3

print(x < y) #False
print(x > y) #True
print(x <= y) #False
print(x >= y) #True
print(x == y) #False
print(x != y) #True

x = 3.0
y = 3

print(x == y) #True
```

## conditional statements

if-statement

0 --> even getal

```python
getal = int(input("Geef een getal in: ")) #3

if getal > 0:
    print("POSITIEF")
elif getal < 10:
    print("KLEINER")
elif getal == 3:
    print("3333333")
else:
    print("TEST")


if getal > 0:
    print(getal, "POSITIEF")
elif getal < 0:
    print(getal, "NEGATIEF")
else:
    print(getal, "GETAL IS 0")



if getal == 0:
    print(getal, "is 0")
else:
    if getal < 0:
        print("negatief")
    else:
        print("positief")


#schrijf een if-statement dat kan checken of een gegeven getal even of oneven is.
if getal % 2 == 0:
    print("even")
else:
    print("oneven")
```
