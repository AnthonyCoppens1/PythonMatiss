Datatypes?

- bool --> boolean --> True en False

```python
#input vragen aan iets of iemand?
#mini opdracht: zoek het datatype van een bepaald getal
x = input()
print(x)
print(f"Het gegeven getal is: {x}")
print("Het gegeven getal is:", x)
print("Het gegeven getal is:" + x)

print(f"Het datatype van x is nu {type(x)}")

#converten van het ene datatype naar het andere
#van str naar int
xAlsInt = int(x)
print(type(xAlsInt))
print(xAlsInt)

xAlsFloat = float(x)
print(type(xAlsFloat))
print(xAlsFloat)

xAlsString = str(xAlsFloat)
print(xAlsString)

```

## Concatenatie en combineren van verschillende outputs
- '+' --> combineren van 2 STRINGS of het optellen van 2 getallen, je moet zelf een spatie vorzien als het een string is en alles blijft een string.
- ',' --> voegt automatisch zelf een spatie toe en laat je toe meerdere datatypes te combineren.

```python
getal = 5
print("Het getal is", getal)
print(getal + getal)
print("Het getal is " + str(getal))

print(str(getal) + str(getal)) # "5" + "5"
print(str(getal + getal))

#print(str(getal) + getal) --> strings en getallen niet op te tellen
```

## Oefeningen les 1 - if-statements
- Schrijf een programma met een if-statement dat checkt of een persoon oud genoeg is om te mogen gaan stemmen.
MAAR let op: het moet gebaseerd zijn op het geboortejaar EN je moet werken met datetime.
De gebruiker geeft zelf het geboortejaar in en je berekent dan de leeftijd van de persoon adhv datetime en geboortejaar
2 outputs: "MAG stemmen" of "mag NIET stemmen"

- schrijf een programma dat checkt of een string leeg is adhv een if-statement. 2 outputs: "string is leeg" OF "string is niet leeg"

- schrijf een programma dat checkt adhv een if-statement of een wachtwoord veilig is.
Veilig ww: minstense 8 characters EN zowel letters als cijfers
2 outputs: "wachtwoord is VEILIG" OF "wachtwoord NIET VEILIG"

```python
#OEF 1
from datetime import datetime as dt

geboortejaar = input("Geef je geboortejaar in: ")
huidigeJaar = dt.now().year

leeftijd = huidigeJaar - int(geboortejaar)
print(leeftijd)
if leeftijd >= 18:
    print("Jeej je mag stemmen")
else:
    print("Woops, te jong om te stemmen")
```

```python
#OEF 2
stringOmTeChecken = input()

#optie 1
if stringOmTeChecken == "":
    print("string is leeg")
else:
    print("string is niet leeg")

#optie 2
if len(stringOmTeChecken) == 0:
    print("string is leeg")
else:
    print("string is niet leeg")

#optie 3
if not stringOmTeChecken:
    print("string is leeg")
else:
    print("string is niet leeg")
```

```python
#OEF 3
tekstOmTeChecken = input("Geef een wachtwoord: ")

if len(tekstOmTeChecken) >= 8 and any(char.isdigit() for char in tekstOmTeChecken) and any(char.isalpha() for char in tekstOmTeChecken):
    print("wachtwoord veilig")
else:
    print("wachtwoord niet veilig")
```

## Loops

- <b> for-loops </b>: itereren over een bepaalde sequentie of een bepaalde range. Je weet zelf exacxt hoeveel keer je iets herhaalt.
- **while-loop**: zolang je aan een voorwaarde voldoet blijft je iets herhalen.

```python
#sequence van characters of woorden

for c in "een woord":
    print(c)

for c in "een woord":
    print(c, end=" ") #voegt een spatie toe op einde van ieder character ipv een nieuwe lijn als default
```

```python
#for loop met range van bv getallen
for x in range(-5,5):
    print(x)


for x in range(0,10):
    print(x, end = " ")


getal = int(input())
for x in range(0, 11):
    print(f"{getal} * {x} = {x * getal}")
```

```python
woord = input("Geef een woord in, zeg STOP om te stoppen")

while woord != "STOP":
    print(woord)
    print("Matiss leert python")
    woord = input()

print("programma is gestopt")

x = 5
while x <= 10000:
    x *= 5
    print(x, end = " ")

print()
x = 5
while x <= 10000:
    print(x, end = " ")
    x *= 5

```
